```python
import pandas as pd
import csv
import re
import os
import glob
```

```python
# Duplicate check in curated data

csvs  = glob.glob('curated_data/**.csv')
pass_lst = []
fail_lst = []

csvs = ['curated_data/player_data_dim.csv']

for file in csvs:
    df = pd.read_csv(file)
    dupes = df[df.duplicated()]

    print(dupes)

    if dupes.empty:
        pass_lst.append(file)
    else:
        fail_lst.append(file)

print(f'List of CSVs with no dupes: {pass_lst}')
print(f'List of CSVs with dupes: {fail_lst}')
```
